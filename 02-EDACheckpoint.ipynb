## Data Cleaning

We first loaded the animal intake outcome dataset and reviewed the fields it contained. Since not all columns were relevant to our study, we selected only the variables that would help us understand patterns in adoption outcomes—such as basic animal characteristics, intake circumstances, and outcome details. This allowed us to focus the dataset on information directly related to our research question and avoid unnecessary columns like `geographic coordinates`, `system flags`, or `internal tracking fields`.

We removed the Outcome Subtype field because it represents organizational or processing pathways (such as partner rescues, walk-in adoptions, or online listings), which do not reflect intrinsic characteristics of the animals themselves. Since our analysis focuses specifically on how animal traits influence adoption outcomes, this field was not relevant and introduced unnecessary categorical noise.

We then converted date fields (DOB, Intake Date, Outcome Date) from strings to datetime objects, preparing them for later analysis. Since our project focuses on adoption patterns specifically for cats and dogs, we filtered the dataset to include only those two species.


```python
df.columns
```

```python
# Select only the useful columns
useful_cols = [
    'Animal ID', 'Animal Type', 'Primary Color', 'Secondary Color',
    'Sex', 'DOB', 'Intake Date', 'Intake Condition', 'Intake Type',
    'Reason for Intake', 'Outcome Date', 'Outcome Type'
]

df = df[useful_cols]

# Remove rows missing essential information
essential_cols = ['Animal Type', 'Outcome Type', 'Sex', 'Intake Date', 'Outcome Date']
df = df.dropna(subset=essential_cols)

#Convert dates
df['DOB'] = pd.to_datetime(df['DOB'], errors='coerce')
df['Intake Date'] = pd.to_datetime(df['Intake Date'], errors='coerce')
df['Outcome Date'] = pd.to_datetime(df['Outcome Date'], errors='coerce')

# Keep only cats and dogs
df = df[df['Animal Type'].str.strip().str.upper().isin(['DOG', 'CAT'])]
```

We then restricted the dataset to cats and dogs, as these species constitute the majority of shelter adoptions and are the primary focus of our project. For outcome analysis, we retained all records except those representing scenarios in which adoption was never possible. Specifically, we removed animals whose outcomes were Euthanasia, Died, Disposal, Missing, or Duplicate, since these cases do not inform adoption behavior and would incorrectly inflate the number of non-adopted animals.

Additionally, we removed all animals whose outcome type was Return to Owner. While these animals were successfully reunited with their original homes, they were never part of the adoptable population, and therefore cannot inform adopter preferences or factors that influence adoption outcomes. Removing return-to-owner cases ensures that our dataset includes only animals who were genuinely eligible for adoption, eliminating a significant source of bias in downstream analyses.

For the remaining animals, we created a binary Adopted variable, coding animals with an outcome type of Adoption as 1 and all other remaining outcomes as 0. This allowed us to model adoption outcomes directly while maintaining a clear distinction between adopted animals and animals who exited the shelter through other non-adoption pathways.


```python
df['Outcome Type'].value_counts()
```

```python
#Drop animals never eligible for adoption
drop_outcomes = ['EUTHANASIA', 'DIED', 'DISPOSAL', 'MISSING', 'DUPLICATE']
df = df[~df['Outcome Type'].str.upper().isin(drop_outcomes)]

#Create “Adopted” indicator
success_outcomes = ['ADOPTION', 'RETURN TO OWNER']
df['Adopted'] = df['Outcome Type'].str.upper().isin(success_outcomes).astype(int)

#Create dataset for adoption modeling only
df_adoption_only = df[df['Outcome Type'].str.upper() != 'RETURN TO OWNER']
```

Finally, we standardized the sex field. The original dataset combined biological sex and sterilization status in a single string (e.g., “Neutered Male”). To improve clarity, we extracted two boolean variables,  `Neutered` and `Spayed`, and standardized the Sex field to contain only `Male` or `Female`. After completing these steps, we reset the index to produce a clean, analysis-ready dataset.

```python
df['Neutered'] = df['Sex'].str.contains('Neutered', case=False, na=False)
df['Spayed'] = df['Sex'].str.contains('Spayed', case=False, na=False)

df['Sex'] = df['Sex'].str.replace('Neutered', 'Male', case=False)
df['Sex'] = df['Sex'].str.replace('Spayed', 'Female', case=False)
df['Sex'] = df['Sex'].str.strip().str.title()

df = df.reset_index(drop=True)
```

### 2. Species Differences (Cats vs. Dogs)

Species is one of the most basic and influential attributes in shelters.
Dogs typically have higher adoption rates than cats, so comparing adoption outcomes by species provides an important baseline understanding of population differences.


```python
sns.countplot(data=df, x='Animal Type', hue='Adopted')
plt.title("Adoption Outcome by Species")
```

### 3. Age Distribution and Its Relationship to Adoption

We created the `age` column base on the intake Date - DOB, as age is a strong predictor of adoption.

```python
df['Age'] = (df['Intake Date'] - df['DOB']).dt.days / 365.25
df = df[(df['Age'] >= 0) & (df['Age'] < 40)]

sns.histplot(df['Age'], bins=30)
plt.title("Distribution of Animal Ages at Intake")
```

```python
sns.boxplot(data=df, x='Adopted', y='Age')
plt.title("Age Differences Between Adopted and Non-Adopted Animals")
plt.xticks([0, 1], ['Not Adopted', 'Adopted'])
```

### 4. Sex Differences (Male vs. Female)
Some studies suggest sex may influence adoptability due to behavioral stereotypes (e.g., male cats perceived as more affectionate). Comparing adoption rates by sex helps evaluate whether such trends appear in this dataset.

```python
sns.countplot(data=df, x='Sex', hue='Adopted')
plt.title("Adoption Outcome by Sex")
```

### 5. Sterilization Status (Spayed/Neutered Combined)
Both spaying and neutering represent the same concept: the animal is sterilized. Since shelters typically require sterilization prior to adoption, this combined feature allows us to check whether sterilized animals have higher adoption rates.

```python
df['Sterilized'] = df['Neutered'] | df['Spayed']

sns.countplot(data=df, x='Sterilized', hue='Adopted')
plt.title("Adoption Outcome by Sterilization Status")
plt.xticks([0, 1], ['Not Sterilized', 'Sterilized'])
```

### 6. Color and Adoption Rates
Color can affect perceived attractiveness.
But since color has many categories, we only look at the most common ones.

```python
top_colors = df['Primary Color'].value_counts().head(8).index
df_color = df[df['Primary Color'].isin(top_colors)]

sns.countplot(data=df_color, x='Primary Color', hue='Adopted')
plt.title("Adoption Outcome by Primary Color (Top Colors)")
plt.xticks(rotation=45)
```
