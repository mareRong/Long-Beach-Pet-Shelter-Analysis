## Research Question

How do animal characteristics and listing features influence the likelihood and speed of adoption for cats and dogs in the San Diego region?

we will analyze features such as species type, age, gender, size, breed, color, spay/neuter status, and behavioral/environmental variables (e.g. house trained, good with dogs/cats/children). Listing level factors such as distance, number of photos, and publication date will also be examined. We will apply exploratory data analysis and predictive modeling to determine which characteristics most strongly predict faster adoption among cats and dogs in San Diego.


## Background and Prior Work



Throughout the United States, animal shelters are full of pets that need homes. Shelters have pets of many different species, age, and other traits. For most people, many factors go into their decision of which pet to adopt. The traits of the pet are believed to be a main factor, but there could also be additional external factors not related to the traits of the animal.

To many people adopting a pet from a shelter is a significant event. People doing this are not only sourcing a pet in a humane way, but they are also picking a companion that will be with them for a long time. Because of this, most people carefully pick their pet. To the common person, the behavior and the physical features of a pet are reasonable to assume as the main influence in deciding what pet to adopt, but we felt that the main factors influencing pet adoption speed should be investigated, since these two factors are too simple of an answer.

There has been some research into what causes some pets to get adopted quicker than others, for example, the NIH researched factors influencing cat adoption outcomes at animal shelters. In their study, they mentioned how many studies and people come to believe that the physical characteristics of cats are the main determining factor in adoption speed. However, they found that the studies that researched this type of connection have varying outcomes, so they determined local conditions are also a main factor in adoption speed.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1)

Additionally, another source we read found that male exotic cats had some of the shortest times of stay at animal shelters.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) This source provides an interesting comparison for us because it directly supports the notion that animals, cats in particular, are adopted based on their ‘looks.’ In their research, scientists investigated the risk factors affecting the length of stay at the Guelph Humane Society from 2011–2016. This article will provide an interesting source to compare our findings to since it is an example of the ‘simple answer.’

Another source that supports the ‘simple answer’ is an article from Purdue University Extension. In the article, they directly mention how age of the animal (in this source, dogs are the focus), its behavior, and its breed were main factors for people to adopt a dog.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) This source will provide an interesting comparison for our findings, as well as helping us see whether the NIH publication is correct in its claim, or whether they are wrong.

Overall, from our findings, we hope to further research what determines how quickly a shelter pet is adopted, and whether it is mainly influenced by external factors or if pets are simply adopted faster for their looks and behavior.

## Hypothesis


We hypothesize that both a shelter pet’s age and the shelter’s distance from San Diego have a positive correlation with its adoption speed. Here, we define “adoption speed” as the number of days between the intake date and the adoption date. In other words, younger animals are expected to be adopted more quickly, and pets located in shelters farther from adopters' addresses are expected to experience a longer waiting period before adoption. We believe that younger pets attract more adopters due to their perceived trainability, and that greater travel distance could act as a logistical barrier that discourages adopters from adopting.

We also hypothesize that the amount of house training and health status, such as the number of vaccines or shots received, are negatively correlated with adoption speed, meaning that pets with more training and vaccinations are likely to have a shorter waiting period before adoption.
