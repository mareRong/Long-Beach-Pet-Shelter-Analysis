To many people adopting a pet from a shelter is a significant event. People doing this are not only sourcing a pet in a humane way, but they are also picking a companion that will be with them for a long time. Because of this, most people carefully pick their pet. # Hypothesis


Using the Long Beach Animal Shelter Intakes and outcomes and dataset, We hypothesize that both a shelter pet’s age and the shelter’s distance from San Diego have a positive correlation with its adoption speed. Here, we define “adoption speed” as the number of days between the intake date and the adoption date. In other words, younger animals are expected to be adopted more quickly, and pets located in shelters farther from adopters' addresses are expected to experience a longer waiting period before adoption. We believe that younger pets attract more adopters due to their perceived trainability, and that greater travel distance could act as a logistical barrier that discourages adopters from adopting.

We also hypothesize that the amount of house training and health status, such as the number of vaccines or shots received, are negatively correlated with adoption speed, meaning that pets with more training and vaccinations are likely to have a shorter waiting period before adoption.

Finally, we expect that spayed or neutred animals will be adopted quickly more than intact one. As they pose fewer veteinary cost, and are often viewed as more suitable for immediate placement.
